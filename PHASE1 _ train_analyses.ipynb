```python
# Target distribution
print(df["Will_Buy_EV"].value_counts())
print(df["Will_Buy_EV"].value_counts(normalize=True))

# Target rate by categorical variables
categorical_cols = [
    "Gender",
    "City_Type",
    "Current_Car_Type",
    "Home_Charging_Possible",
    "Subsidy_Available",
    "Range_Anxiety_Level"
]

for col in categorical_cols:
    print("\n" + "="*60)
    print(col)
    print(
        df.groupby(col)["Will_Buy_EV"]
        .value_counts(normalize=True)
        .unstack()
        .round(4)
    )
```

```python
# Numerical feature correlations with target
target_num = df["Will_Buy_EV"].map({"No": 0, "Yes": 1})

numeric_cols = [
    "Age",
    "Annual_Income_USD",
    "Daily_Commute_km",
    "Number_of_Cars_Owned",
    "Charging_Stations_Near_Home",
    "Charging_Stations_Near_Work",
    "Environmental_Concern_Level"
]

print(
    df[numeric_cols]
    .corrwith(target_num)
    .sort_values(ascending=False)
)
```

```python
df["Concern_x_Subsidy"] = ( df["Environmental_Concern_Level"] * (df["Subsidy_Available"] == "Yes").astype(int) )
```

```python
triple = (
    df.groupby(
        [
            "Subsidy_Available",
            "Range_Anxiety_Level",
            "Environmental_Concern_Level"
        ]
    )["Will_Buy_EV"]
    .apply(lambda x: (x == "Yes").mean())
    .reset_index(name="purchase_rate")
)

triple
```

```python
analysis = (
    df.groupby(
        [
            "Subsidy_Available",
            "Range_Anxiety_Level",
            "Environmental_Concern_Level"
        ]
    )
    .agg(
        count=("Will_Buy_EV", "size"),
        purchase_rate=(
            "Will_Buy_EV",
            lambda x: (x == "Yes").mean()
        )
    )
    .reset_index()
)

print(analysis.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df["target"] = (df["Will_Buy_EV"] == "Yes").astype(int)
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Will_Buy_EV"
)

plt.title("EV Purchase Distribution")
plt.xlabel("Will Buy EV")
plt.ylabel("Number of Customers")

plt.show()
```

```python
env_rate = (
    df.groupby("Environmental_Concern_Level")["target"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=env_rate,
    x="Environmental_Concern_Level",
    y="target"
)

plt.title("EV Purchase Rate by Environmental Concern")
plt.xlabel("Environmental Concern Level")
plt.ylabel("EV Purchase Rate")

plt.ylim(0, 1)

plt.show()
```

```python
subsidy_env = (
    df.groupby(
        ["Environmental_Concern_Level", "Subsidy_Available"]
    )["target"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(9, 5))

sns.lineplot(
    data=subsidy_env,
    x="Environmental_Concern_Level",
    y="target",
    hue="Subsidy_Available",
    marker="o"
)

plt.title("EV Purchase Rate: Subsidy × Environmental Concern")
plt.xlabel("Environmental Concern Level")
plt.ylabel("EV Purchase Rate")

plt.ylim(0, 1)

plt.show()
```

```python
anxiety_env = (
    df.groupby(
        ["Environmental_Concern_Level", "Range_Anxiety_Level"]
    )["target"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(9, 5))

sns.lineplot(
    data=anxiety_env,
    x="Environmental_Concern_Level",
    y="target",
    hue="Range_Anxiety_Level",
    marker="o"
)

plt.title("EV Purchase Rate: Range Anxiety × Environmental Concern")
plt.xlabel("Environmental Concern Level")
plt.ylabel("EV Purchase Rate")

plt.ylim(0, 1)

plt.show()
```

```python
# 3-way interaction with counts + actual number of buyers
analysis = (
    df.groupby(
        [
            "Subsidy_Available",
            "Range_Anxiety_Level",
            "Environmental_Concern_Level"
        ]
    )
    .agg(
        count=("Will_Buy_EV", "size"),
        buyers=("target", "sum"),
        purchase_rate=("target", "mean")
    )
    .reset_index()
)

print(analysis.to_string(index=False))
```

```python
# Income effect inside the strongest group
strong_group = df[
    (df["Subsidy_Available"] == "Yes") &
    (df["Range_Anxiety_Level"] == "Low") &
    (df["Environmental_Concern_Level"] == 5)
].copy()

strong_group["income_bin"] = pd.qcut(
    strong_group["Annual_Income_USD"],
    10,
    duplicates="drop"
)

print(
    strong_group.groupby("income_bin", observed=True)["target"]
    .agg(["count", "mean"])
)
```
